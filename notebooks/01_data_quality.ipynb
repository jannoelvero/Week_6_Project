```python
#Count starts and confirms per client session
#Purpose: This creates one row for each client session and counts how many times each process step occurred.
session_steps = (
    web_clean_df
    .groupby(["client_id", "visit_id", "process_step"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

session_steps.head()
```

```python
#Investigate multiple starts
multiple_starts = session_steps[
    session_steps["start"] > 1
]

print(
    "Sessions with multiple starts:",
    len(multiple_starts)
)
```

```python
multiple_starts.head(10)
```

```python
#Investigate multiple confirms
multiple_confirms = session_steps[
    session_steps["confirm"] > 1
]

print(
    "Sessions with multiple confirms:",
    len(multiple_confirms)
)
```

```python
multiple_confirms.head(10)
```

```python
#Sessions without a start
# if a session without a recorded start cannot straightforwardly be used to calculate full journey completion time.
no_start = session_steps[
    session_steps["start"] == 0
]

print(
    "Sessions without a start:",
    len(no_start)
)
```

```python
#Sessions without a confirm
#may contain abandoned or incomplete journeys
no_confirm = session_steps[
    session_steps["confirm"] == 0
]

print(
    "Sessions without a confirm:",
    len(no_confirm)
)
```

```python
### Session Event Quality Investigation

#After removing exact duplicate events, the Digital Footprints dataset contains 744,641 recorded events.

#Customer sessions were defined using the combination of `client_id` and `visit_id` because the earlier identifier investigation showed that some `visit_id` values are associated with multiple clients.

#The session-level investigation identified substantial variation in customer navigation behaviour.

#A total of 49,873 sessions contain multiple `start` events, while 8,248 sessions contain multiple `confirm` events. These events will not automatically be removed because they may represent genuine customer behaviour such as restarting, repeating actions, or navigating through the process multiple times within a session.

#There are also 13,277 sessions without a recorded `start`. These sessions cannot be used straightforwardly to calculate full journey completion time because the beginning of the journey is not observed.

#A total of 68,474 sessions contain no `confirm` event. These sessions will not be removed because they may represent incomplete or abandoned journeys, which are directly relevant to the analysis of process completion.

#The findings demonstrate that a browsing session cannot automatically be treated as a single linear customer journey. A consistent journey reconstruction methodology must therefore be established before calculating completion, duration, navigation error, and funnel KPIs.
```

```python
### Session Pattern Summary

#The cleaned Digital Footprints dataset contains 159,112 client-session combinations.

#Among these sessions, 31.34% contain more than one `start` event, while 5.18% contain multiple `confirm` events. In addition, 8.34% of sessions have no recorded `start`, and 43.04% have no recorded `confirm`.

#These categories are not mutually exclusive. A single session may belong to more than one category.

#The high frequency of multiple starts and incomplete sessions indicates that customer behaviour is often non-linear. These patterns should therefore not be treated automatically as data errors.

#Sessions without a `confirm` will be retained because incomplete journeys are directly relevant to Vanguard's completion and abandonment questions.

#Sessions without a `start` will also be retained for navigation analysis, although they cannot be used straightforwardly for full start-to-confirm completion time calculations.

#These findings confirm that customer journeys must be reconstructed carefully before business KPIs are calculated.
```

```python
###backward navigation
#to check whether clients experience fewer navigation errors.
step_order = {
    "start": 0,
    "step_1": 1,
    "step_2": 2,
    "step_3": 3,
    "confirm": 4
}
web_clean_df["step_number"] = (
    web_clean_df["process_step"]
    .map(step_order)
)
web_clean_df = web_clean_df.sort_values(
    ["client_id", "visit_id", "date_time"]
)
web_clean_df["previous_step"] = (
    web_clean_df
    .groupby(["client_id", "visit_id"])["step_number"]
    .shift(1)
)
web_clean_df["backward_navigation"] = (
    web_clean_df["step_number"]
    < web_clean_df["previous_step"]
)

print(
    "Backward navigation events:",
    web_clean_df["backward_navigation"].sum()
)
```

```python
#62,880 backward navigation events after exact duplicates were removed.
#does not mean 62,880 clients made errors. One client can generate several backward events, and one session can contain several.
```

```python
#How many sessions experienced backward navigation?
```

```python
sessions_with_backward = (
    web_clean_df
    .groupby(["client_id", "visit_id"])["backward_navigation"]
    .any()
    .sum()
)

print(
    "Sessions with backward navigation:",
    sessions_with_backward
)

print(
    "Percentage of sessions with backward navigation:",
    round(sessions_with_backward / total_sessions * 100, 2),
    "%"
)
```

```python
#Which backward movements occur most often? to reveal where customers struggle.

This can reveal where customers struggle.

backward_events = web_clean_df[
    web_clean_df["backward_navigation"] == True
].copy()

backward_events["previous_process_step"] = (
    backward_events["previous_step"]
    .map({
        0: "start",
        1: "step_1",
        2: "step_2",
        3: "step_3",
        4: "confirm"
    })
)

backward_events[
    ["previous_process_step", "process_step"]
].value_counts()
```

```python
### Backward Navigation Findings

#A total of 62,880 backward navigation events were identified across 39,949 client sessions after exact duplicate events were removed.

#This means that 25.11% of the 159,112 client sessions experienced at least one backward movement.

#The most common backward transition was `step_1` to `start`, with 23,587 occurrences. Other frequent backward transitions included `step_2` to `step_1` with 12,171 occurrences, `step_3` to `step_2` with 9,418 occurrences, and `step_3` to `start` with 8,265 occurrences.

#Backward navigation is not automatically classified as an error because customers may legitimately return to previous steps to review or correct information. However, its frequency provides an important indicator of potential navigation friction.

#The Test and Control groups will later be compared to determine whether the redesigned interface reduced backward navigation and whether particular process steps remain potential bottlenecks.
```
